```python
# 1. import pandas and scipy
import scipy.stats as STATS
import pandas as pd

# 2. Load the cleaned dataset
df = pd.read_csv("GSE306399_matrix_transcript.fpkm.csv")
df = df.drop(columns=['Unnamed: 1'], errors='ignore')

# Separate the two treatment groups to compare
# KD (Knockdown) vs SC (Control)
group_kd = df[['KD-veh1', 'KD-veh2', 'KD-veh3']]
group_sc = df[['SC-veh1', 'SC-veh2', 'SC-veh3']]

# 3. Perform a two-sample independent T-test to compare the conditions
# We apply the t-test row-wise (axis=1) across transcripts
STAT_RESULT, P_VALUE = STATS.ttest_ind(group_kd, group_sc, axis=1)

# Calculate the number of significantly different transcripts
p_val_col = pd.Series(P_VALUE).dropna()
significant_count = (p_val_col < 0.05).sum()

# 4. Print results
print(f"Number of analyzed transcripts: {len(p_val_col)}")
print(f"Number of transcripts with a significant difference (P < 0.05): {significant_count}")

# Add the p-value column to the dataframe and display the first 10 rows
df['p_value'] = P_VALUE
display(df[['transcript_id', 'p_value']].head(10))
```

```python
# 1. Calculate the arithmetic mean for each group
mean_KD = df[['KD-veh1', 'KD-veh2', 'KD-veh3']].mean(axis=1)
mean_SC = df[['SC-veh1', 'SC-veh2', 'SC-veh3']].mean(axis=1)

# 2. Calculate Fold Change (adding a small epsilon of 0.001 to prevent division by zero)
df['fold_change'] = (mean_KD + 0.001) / (mean_SC + 0.001)

# 3. Determine the direction of regulation (Upregulated or Downregulated)
import numpy as np
df['log2_FC'] = np.log2(df['fold_change'])

# 4. Classify genes with statistically significant differences (P < 0.05)
significant = df[df['p_value'] < 0.05]

upregulated = significant[significant['log2_FC'] > 0]
downregulated = significant[significant['log2_FC'] < 0]

print("Number of Upregulated genes:", len(upregulated))
print("Number of Downregulated genes:", len(downregulated))
```

```python
import numpy as np

# 1. Calculate means and Fold Change
mean_KD = df[['KD-veh1', 'KD-veh2', 'KD-veh3']].mean(axis=1)
mean_SC = df[['SC-veh1', 'SC-veh2', 'SC-veh3']].mean(axis=1)

df['fold_change'] = (mean_KD + 0.001) / (mean_SC + 0.001)
df['log2_FC'] = np.log2(df['fold_change'])

# 2. Filter for significant genes only (P-value < 0.05)
significant_genes = df[df['p_value'] < 0.05].copy()

# 3. Sort upregulated genes from highest to lowest fold change (Top Upregulated)
top_upregulated = significant_genes.sort_values(by='log2_FC', ascending=False)

# 4. Sort downregulated genes (Top Downregulated)
top_downregulated = significant_genes.sort_values(by='log2_FC', ascending=True)

# 5. Display top 10 upregulated genes
print("=== TOP UPREGULATED GENES ===")
display(top_upregulated[['transcript_id', 'log2_FC', 'p_value']].head(10))

# 6. Display top 10 downregulated genes
print("=== TOP DOWNREGULATED GENES ===")
display(top_downregulated[['transcript_id', 'log2_FC', 'p_value']].head(10))
```

```python
import numpy as np
import plotly.express as px

# 1. Recalculate means to ensure values are present
mean_KD = df[['KD-veh1', 'KD-veh2', 'KD-veh3']].mean(axis=1)
mean_SC = df[['SC-veh1', 'SC-veh2', 'SC-veh3']].mean(axis=1)

# 2. Calculate Fold Change and log2_FC
df['fold_change'] = (mean_KD + 0.001) / (mean_SC + 0.001)
df['log2_FC'] = np.log2(df['fold_change'])

# 3. Calculate negative log10 p-value for the plot
df['-log10_pvalue'] = -np.log10(df['p_value'] + 1e-300)

# 4. Classify genes (Upregulated vs Downregulated)
conditions = [
    (df['p_value'] < 0.05) & (df['log2_FC'] > 1),
    (df['p_value'] < 0.05) & (df['log2_FC'] < -1)
]
choices = ['Upregulated', 'Downregulated']
df['Status'] = np.select(conditions, choices, default='Not Significant')

# 5. Generate the interactive Volcano Plot using Plotly
fig = px.scatter(
    df,
    x='log2_FC',
    y='-log10_pvalue',
    hover_name='transcript_id',  # Shows the transcript ID on hover
    color='Status',
    color_discrete_map={
        'Upregulated': 'green',
        'Downregulated': 'red',
        'Not Significant': 'lightgrey'
    },
    title='Interactive Volcano Plot - Effect of PD-L1 Knockdown'
)

fig.show()
```
